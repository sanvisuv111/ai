```python
class Node:
    def __init__(self, data, level, fval):
        self.data = data
        self.level = level
        self.fval = fval

    def generate_child(self):
        x, y = self.find(self.data, '_')
        # Directions: Left, Right, Up, Down
        val_list = [[x, y-1], [x, y+1], [x-1, y], [x+1, y]]
        children = []
        for i in val_list:
            child = self.shuffle(self.data, x, y, i[0], i[1])
            if child is not None:
                child_node = Node(child, self.level + 1, 0)
                children.append(child_node)
        return children

    def shuffle(self, puz, x1, y1, x2, y2):
        if 0 <= x2 < len(self.data) and 0 <= y2 < len(self.data):
            temp_puz = self.copy(puz)
            temp_puz[x1][y1], temp_puz[x2][y2] = temp_puz[x2][y2], temp_puz[x1][y1]
            return temp_puz
        else:
            return None

    def copy(self, root):
        return [row[:] for row in root]

    def find(self, puz, x):
        for i in range(len(puz)):
            for j in range(len(puz[i])):
                if puz[i][j] == x:
                    return i, j


class Puzzle:
    def __init__(self, size):
        self.n = size
        self.open = []
        self.closed = []

    def accept(self, msg):
        print(msg)
        puz = []
        for i in range(self.n):
            row = input(f"Enter row {i+1} (space-separated): ").split()
            puz.append(row)
        return puz

    def f(self, start, goal):
        return self.h(start.data, goal) + start.level

    def h(self, start, goal):
        count = 0
        for i in range(self.n):
            for j in range(self.n):
                if start[i][j] != goal[i][j] and start[i][j] != '_':
                    count += 1
        return count

    def process(self):
        start = self.accept("Enter the START state matrix:")
        goal = self.accept("Enter the GOAL state matrix:")

        start_node = Node(start, 0, 0)
        start_node.fval = self.f(start_node, goal)
        self.open.append(start_node)

        while True:
            if not self.open:
                print("No solution found!")
                return

            current = self.open.pop(0)
            print("\nCurrent state (fval = {}):".format(current.fval))
            for row in current.data:
                print(" ".join(row))

            if self.h(current.data, goal) == 0:
                print("\nGoal state reached!")
                return

            children = current.generate_child()
            for child in children:
                if not any(child.data == node.data for node in self.open + self.closed):
                    child.fval = self.f(child, goal)
                    self.open.append(child)

            self.closed.append(current)
            self.open.sort(key=lambda x: x.fval)


# Run the solver
if __name__ == "__main__":
    puz = Puzzle(3)
    puz.process()
```
